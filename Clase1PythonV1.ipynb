## Comparación: Datos Sueltos vs Datos Estructurados

### Observa la diferencia entre estos dos enfoques

En el trabajo con APIs, es fundamental entender cómo se organizan los datos. Compara los siguientes dos ejemplos:

````python
# ============================================
# EJEMPLO A: Datos sueltos (Desordenados)
# ============================================

vin = "W1N12345"
temp = 95
alerta = True

print("=== ENFOQUE A: VARIABLES SUELTAS ===")
print("VIN:", vin)
print("Temperatura:", temp)
print("Alerta:", alerta)

# ============================================
# EJEMPLO B: Dato estructurado (Como llega en una API real)
# ============================================

telemetria_unidad = {
    "chasis": "W1N12345",
    "motor": {
        "temperatura": 95,
        "presion_aceite": 2.1
    },
    "estado_critico": True
}

print("\n=== ENFOQUE B: DATO ESTRUCTURADO ===")
print("Datos completos:", telemetria_unidad)
print("Solo el chasis:", telemetria_unidad["chasis"])
print("Temperatura del motor:", telemetria_unidad["motor"]["temperatura"])
````

### Análisis de la Comparación

Observa las diferencias clave entre los dos enfoques:

**Enfoque A - Variables Sueltas:**

- Cada dato tiene su propia variable independiente
- No hay relación explícita entre vin, temp y alerta
- Difícil de pasar como un solo conjunto de datos a una función
- No escala bien cuando tienes muchos vehículos

**Enfoque B - Dato Estructurado:**

- Todos los datos relacionados están agrupados en una sola estructura
- La relación entre los datos es clara y explícita
- Se puede pasar todo el conjunto a una función con un solo parámetro
- Escala bien para múltiples vehículos y datos complejos

**¿Por qué esto importa para las APIs?**

Cuando trabajas con APIs, los datos casi siempre llegan en el formato del Enfoque B. Las APIs modernas usan JSON, que en Python se representa exactamente como diccionarios anidados.

Entender cómo navegar por estas estructuras es una habilidad fundamental que desarrollarás en este curso.

## Concepto 1: Variables - Los Contenedores de Datos

### ¿Qué es una variable?

Una variable es como una caja con una etiqueta donde guardamos información para usarla después. En programación, las variables nos permiten almacenar valores y referirnos a ellos por un nombre.

### Tipos de Datos Básicos en Python

Python maneja varios tipos de datos fundamentales. Los cuatro más importantes son:

**1. String (Texto)**

Se usa para almacenar texto. Siempre va entre comillas (simples o dobles).

Ejemplo: nombre = "Mercedes"

**2. Integer (Número Entero)**

Se usa para números sin decimales.

Ejemplo: anio = 2024

**3. Float (Número Decimal)**

Se usa para números con decimales.

Ejemplo: precio = 45000.50

**4. Boolean (Verdadero/Falso)**

Se usa para valores de verdad. Solo puede ser True o False.

Ejemplo: disponible = True

### Reglas para nombrar variables

- Los nombres deben ser descriptivos (ej: precio_vehiculo en lugar de p)
- No pueden comenzar con números
- No pueden contener espacios (usa guion bajo: precio_vehiculo)
- Distinguen entre mayúsculas y minúsculas (Precio es diferente a precio)

````python
# ============================================
# EJEMPLO 1: DECLARACIÓN DE VARIABLES
# ============================================

# Variables de tipo String (texto)
marca = "Mercedes"
modelo = "Clase C"

# Variables de tipo Integer (enteros)
anio = 2024
kilometros = 15000

# Variables de tipo Float (decimales)
precio = 45000.50
consumo = 7.5

# Variables de tipo Boolean (verdadero/falso)
disponible = True
en_garantia = False

# ============================================
# MOSTRAR VALORES EN PANTALLA
# ============================================

print("Marca:", marca)
print("Modelo:", modelo)
print("Año:", anio)
print("Precio:", precio)
print("Disponible:", disponible)

# ============================================
# VERIFICAR TIPOS DE DATOS
# ============================================

print("\n=== TIPOS DE DATOS ===")
print("Tipo de marca:", type(marca))
print("Tipo de anio:", type(anio))
print("Tipo de precio:", type(precio))
print("Tipo de disponible:", type(disponible))
````

### Análisis del Código Anterior

Observa los siguientes puntos en el ejemplo anterior:

**1. El símbolo de asignación (=)**

El signo igual no significa "igualdad" como en matemáticas. Significa "asignar". Cuando escribes:

precio = 45000.50

Le estás diciendo a Python: "guarda el valor 45000.50 en una caja llamada precio"

**2. La función print()**

La función print() muestra información en pantalla. Es la forma más básica de ver qué valor tiene una variable.

**3. Las comas en print()**

Cuando usas comas dentro de print(), Python separa automáticamente los elementos con un espacio.

print("Precio:", precio)

Muestra: Precio: 45000.5

**4. Los comentarios (#)**

Las líneas que comienzan con # son comentarios. Python las ignora al ejecutar el código. Sirven para que los humanos entiendan qué hace cada parte del script.

**5. La función type()**

La función type() te dice qué tipo de dato es una variable. Esto es útil para debugging cuando no estás seguro de qué tipo de dato estás manejando.

## Concepto 2: Listas - Colecciones Ordenadas

### ¿Qué es una lista?

Una lista es una colección ordenada de valores. Piensa en ella como una columna de Excel donde cada fila tiene un valor.

**Características de las listas:**

- Se definen con corchetes []
- Los elementos se separan por comas
- Cada elemento tiene una posición numérica (índice) que comienza en 0
- Pueden contener cualquier tipo de dato
- Pueden modificarse después de creadas (son mutables)

### Operaciones con Listas

**1. Crear una lista**

mi_lista = ["elemento1", "elemento2", "elemento3"]

**2. Acceder a elementos**

mi_lista[0]  # Primer elemento
mi_lista[1]  # Segundo elemento
mi_lista[-1] # Último elemento

**3. Agregar elementos**

mi_lista.append("nuevo_elemento")  # Agrega al final
mi_lista.insert(1, "elemento_en_posicion_1")  # Agrega en posición específica

**4. Modificar elementos**

mi_lista[0] = "valor_modificado"

**5. Remover elementos**

mi_lista.remove("valor_a_eliminar")  # Remueve por valor
mi_lista.pop()  # Remueve el último elemento
mi_lista.pop(0)  # Remueve el elemento en índice 0
del mi_lista[0]  # Otra forma de eliminar por índice

**6. Obtener longitud**

len(mi_lista)  # Devuelve el número de elementos

````python
# ============================================
# EJEMPLO 2: OPERACIONES CON LISTAS
# ============================================

# Crear una lista inicial
vehiculos = ["Mercedes", "BMW", "Audi"]

print("=== LISTA ORIGINAL ===")
print("Vehículos:", vehiculos)
print("Longitud:", len(vehiculos))

# ============================================
# ACCEDER A ELEMENTOS
# ============================================

print("\n=== ACCESO A ELEMENTOS ===")
print("Primer vehículo:", vehiculos[0])
print("Segundo vehículo:", vehiculos[1])
print("Último vehículo:", vehiculos[-1])

# ============================================
# AGREGAR ELEMENTOS
# ============================================

print("\n=== AGREGAR ELEMENTOS ===")
vehiculos.append("Volkswagen")  # Agregar al final
print("Después de append:", vehiculos)

vehiculos.insert(1, "Toyota")  # Agregar en posición 1
print("Después de insert:", vehiculos)

# ============================================
# MODIFICAR ELEMENTOS
# ============================================

print("\n=== MODIFICAR ELEMENTOS ===")
vehiculos[0] = "Mercedes-Benz"  # Modificar primer elemento
print("Después de modificar:", vehiculos)

# ============================================
# REMOVER ELEMENTOS
# ============================================

print("\n=== REMOVER ELEMENTOS ===")
vehiculos.remove("BMW")  # Remover por valor
print("Después de remove:", vehiculos)

ultimo = vehiculos.pop()  # Remover último
print("Después de pop:", vehiculos)
print("Elemento removido:", ultimo)

del vehiculos[0]  # Remover por índice
print("Después de del:", vehiculos)

# ============================================
# OTRAS OPERACIONES ÚTILES
# ============================================

print("\n=== OTRAS OPERACIONES ===")
numeros = [5, 2, 8, 1, 9]
print("Lista original:", numeros)
print("Ordenada:", sorted(numeros))
print("Suma:", sum(numeros))
print("Máximo:", max(numeros))
print("Mínimo:", min(numeros))
````

### Análisis Detallado de las Operaciones

**append() vs insert():**

- append() siempre agrega al final de la lista
- insert() te permite especificar la posición exacta
- insert(0, elemento) agrega al principio de la lista

**remove() vs pop() vs del:**

- remove() busca el valor y elimina la primera coincidencia
- pop() elimina por índice y devuelve el valor eliminado
- del es una declaración que elimina por índice sin devolver el valor

**Índices negativos:**

Python permite usar índices negativos para contar desde el final:

lista[-1] = Último elemento
lista[-2] = Penúltimo elemento
lista[-3] = Antepenúltimo elemento

**Importante para APIs:**

Cuando recibes datos de una API, frecuentemente vienen como listas de objetos. Por ejemplo, una lista de vehículos, una lista de transacciones, o una lista de registros. Saber navegar estas listas es esencial.

## Concepto 3: Diccionarios - Pares Clave-Valor

### ¿Qué es un diccionario?

Un diccionario es una colección de pares clave-valor. Piensa en él como una fila de Excel con encabezados, o como un registro de base de datos.

**Características de los diccionarios:**

- Se definen con llaves {}
- Cada elemento tiene una clave y un valor separados por dos puntos
- Las claves son únicas (no puede haber dos claves iguales)
- Las claves son generalmente strings
- Los valores pueden ser cualquier tipo de dato
- Permiten acceder a los datos por nombre en lugar de por posición

### Sintaxis básica

mi_diccionario = {
    "clave1": "valor1",
    "clave2": "valor2",
    "clave3": "valor3"
}

### Acceder a valores

mi_diccionario["clave1"]  # Devuelve "valor1"

````python
# ============================================
# EJEMPLO 3: OPERACIONES CON DICCIONARIOS
# ============================================

# Crear un diccionario inicial
vehiculo = {
    "marca": "Mercedes",
    "modelo": "Clase C",
    "anio": 2024,
    "precio": 45000.50,
    "disponible": True
}

print("=== DICCIONARIO ORIGINAL ===")
print("Vehículo completo:", vehiculo)

# ============================================
# ACCEDER A VALORES
# ============================================

print("\n=== ACCESO A VALORES ===")
print("Marca:", vehiculo["marca"])
print("Precio:", vehiculo["precio"])

# ============================================
# AGREGAR NUEVAS CLAVES
# ============================================

print("\n=== AGREGAR CLAVES ===")
vehiculo["color"] = "Negro"
print("Después de agregar color:", vehiculo)

vehiculo["kilometros"] = 0
print("Después de agregar kilometros:", vehiculo)

# ============================================
# MODIFICAR VALORES EXISTENTES
# ============================================

print("\n=== MODIFICAR VALORES ===")
vehiculo["precio"] = 43000.00
print("Precio modificado:", vehiculo["precio"])

vehiculo["disponible"] = False
print("Disponibilidad modificada:", vehiculo["disponible"])

# ============================================
# REMOVER CLAVES
# ============================================

print("\n=== REMOVER CLAVES ===")
valor_eliminado = vehiculo.pop("kilometros")
print("Valor eliminado:", valor_eliminado)
print("Diccionario después de pop:", vehiculo)

del vehiculo["color"]
print("Diccionario después de del:", vehiculo)

# ============================================
# MÉTODOS ÚTILES
# ============================================

print("\n=== MÉTODOS ÚTILES ===")
print("Claves:", vehiculo.keys())
print("Valores:", vehiculo.values())
print("Items:", vehiculo.items())
print("Longitud:", len(vehiculo))

# ============================================
# ACCESO SEGURO CON .get()
# ============================================

print("\n=== ACCESO SEGURO ===")
print("Marca:", vehiculo.get("marca"))
print("Color (no existe):", vehiculo.get("color"))
print("Color con valor por defecto:", vehiculo.get("color", "No especificado"))
````

### Análisis Detallado de las Operaciones

**Acceso directo vs .get():**

- diccionario["clave"] lanza un error si la clave no existe
- diccionario.get("clave") devuelve None si la clave no existe
- diccionario.get("clave", "valor_por_defecto") devuelve un valor alternativo

**keys(), values(), items():**

- keys() devuelve todas las claves del diccionario
- values() devuelve todos los valores del diccionario
- items() devuelve pares clave-valor como tuplas

**Importante para APIs:**

Cuando trabajas con APIs, los datos casi siempre vienen en formato de diccionario. El método .get() es especialmente útil porque las APIs no siempre devuelven todos los campos esperados, y .get() evita que tu código se rompa cuando falta un campo.

**Modificar diccionarios:**

Los diccionarios son mutables, lo que significa que puedes cambiar su contenido después de crearlos. Esto es importante cuando necesitas actualizar datos recibidos de una API antes de procesarlos.

## Concepto 4: Diccionarios Anidados

### ¿Qué son los diccionarios anidados?

Un diccionario anidado es un diccionario que contiene otros diccionarios (o listas) como valores. Esta estructura es extremadamente común en las respuestas de APIs.

### ¿Por qué son importantes para las APIs?

Las APIs modernas devuelven datos en formato JSON, que tiene una estructura jerárquica. En Python, JSON se convierte automáticamente en diccionarios anidados.

**Ejemplo de estructura típica de API:**

{
    "estado": "success",
    "datos": {
        "vehiculo": {
            "marca": "Mercedes",
            "motor": {
                "tipo": "V6",
                "cilindrada": 3.0
            }
        }
    }
}

### Cómo acceder a datos anidados

Para acceder a valores profundos, encadenas los corchetes:

respuesta["datos"]["vehiculo"]["motor"]["tipo"]

Esto puede volverse complejo, por lo que es importante practicar la lectura de estas estructuras.

````python
# ============================================
# EJEMPLO 4: DICCIONARIOS ANIDADOS
# ============================================

# Estructura similar a una respuesta de API real
respuesta_api = {
    "estado": "success",
    "codigo": 200,
    "datos": {
        "vehiculo": {
            "vin": "WDD123456789",
            "marca": "Mercedes",
            "modelo": "Clase S",
            "motor": {
                "tipo": "V8",
                "cilindrada": 4.0,
                "potencia": 500
            },
            "sensores": [
                {"nombre": "temperatura", "valor": 95, "unidad": "C"},
                {"nombre": "presion", "valor": 2.5, "unidad": "bar"},
                {"nombre": "nivel_aceite", "valor": 85, "unidad": "%"}
            ]
        },
        "propietario": {
            "nombre": "Juan Pérez",
            "id_cliente": "CLI-001"
        }
    }
}

print("=== ESTRUCTURA COMPLETA ===")
print("Estado:", respuesta_api["estado"])
print("Código:", respuesta_api["codigo"])

print("\n=== ACCESO ANIDADO NIVEL 1 ===")
print("Datos del vehículo:", respuesta_api["datos"]["vehiculo"])

print("\n=== ACCESO ANIDADO NIVEL 2 ===")
print("Marca:", respuesta_api["datos"]["vehiculo"]["marca"])
print("Modelo:", respuesta_api["datos"]["vehiculo"]["modelo"])

print("\n=== ACCESO ANIDADO NIVEL 3 (MOTOR) ===")
print("Tipo de motor:", respuesta_api["datos"]["vehiculo"]["motor"]["tipo"])
print("Cilindrada:", respuesta_api["datos"]["vehiculo"]["motor"]["cilindrada"])
print("Potencia:", respuesta_api["datos"]["vehiculo"]["motor"]["potencia"])

print("\n=== ACCESO A LISTAS DENTRO DE DICCIONARIOS ===")
print("Primer sensor:", respuesta_api["datos"]["vehiculo"]["sensores"][0])
print("Nombre del primer sensor:", respuesta_api["datos"]["vehiculo"]["sensores"][0]["nombre"])
print("Valor del primer sensor:", respuesta_api["datos"]["vehiculo"]["sensores"][0]["valor"])

print("\n=== RECORRER SENSORES ===")
for sensor in respuesta_api["datos"]["vehiculo"]["sensores"]:
    print(f"  {sensor['nombre']}: {sensor['valor']} {sensor['unidad']}")

print("\n=== DATOS DEL PROPIETARIO ===")
print("Nombre:", respuesta_api["datos"]["propietario"]["nombre"])
print("ID Cliente:", respuesta_api["datos"]["propietario"]["id_cliente"])
````

### Análisis de la Estructura Anidada

**Niveles de anidamiento:**

En el ejemplo anterior, podemos identificar varios niveles:

- Nivel 0: respuesta_api (diccionario principal)
- Nivel 1: "estado", "codigo", "datos"
- Nivel 2: "vehiculo", "propietario" (dentro de "datos")
- Nivel 3: "marca", "modelo", "motor", "sensores" (dentro de "vehiculo")
- Nivel 4: "tipo", "cilindrada", "potencia" (dentro de "motor")

**Combinación con listas:**

Observa que "sensores" es una lista que contiene diccionarios. Esto es muy común en APIs que devuelven múltiples registros del mismo tipo.

**Patrón de acceso:**

respuesta_api["datos"]["vehiculo"]["sensores"][0]["nombre"]

Este patrón se lee de izquierda a derecha:
1. Entra en respuesta_api
2. Busca la clave "datos"
3. Dentro de datos, busca "vehiculo"
4. Dentro de vehiculo, busca "sensores"
5. Toma el primer elemento de la lista [0]
6. Dentro de ese elemento, busca "nombre"

### Mejores prácticas para trabajar con anidamiento

**1. Usar .get() para acceso seguro:**

respuesta_api.get("datos", {}).get("vehiculo", {}).get("marca", "Desconocido")

Esto evita errores si alguna clave intermedia no existe.

**2. Extraer referencias intermedias:**

vehiculo = respuesta_api["datos"]["vehiculo"]

marca = vehiculo["marca"]

Esto hace el código más legible cuando necesitas acceder a múltiples valores del mismo nivel.

**3. Validar existencia antes de acceder:**

```python
if "datos" in respuesta_api and "vehiculo" in respuesta_api["datos"]:

    marca = respuesta_api["datos"]["vehiculo"]["marca"]
  ```

| Operación | Método | Ejemplo |
|-----------|--------|---------|
| Agregar al final | append() | lista.append(elemento) |
